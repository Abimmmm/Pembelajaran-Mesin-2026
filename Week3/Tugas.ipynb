```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.model_selection import train_test_split
```

```python
df = pd.read_csv('wbc.csv')
df.info()
```

```python
df.isnull().sum()
```

```python
# 1
# df = df.drop(['id', 'Unnamed: 32'], axis=1)

# X = df.drop('diagnosis', axis=1)
# y = df['diagnosis']

# print("Variabel yang digunakan:")
# print(X.columns.tolist())

X = df.drop('diagnosis', axis=1)
y = df['diagnosis']

print("Variabel yang digunakan:")
print(X.columns.tolist())
```

```python
#  2
df['diagnonis'] = df['diagnosis'].fillna('Unknown')
print("=== Hasil Setelah Fill Nan pada 'diagnosis' ===")
print(df[['diagnosis']].head(33), "\n")
df['diagnosis'] = df['diagnosis'].apply(lambda x: x[0] if x != 'Unknown' else 'U')

le = LabelEncoder()
y = le.fit_transform(y)
print("=== Hasil Setelah Encoding 'diagnosis' ===")
print(dict(zip(le.classes_, le.transform(le.classes_))), "\n")
```

```python
# 3
scaler = StandardScaler()
X = pd.DataFrame(scaler.fit_transform(X), columns=X.columns)
print("=== Hasil Setelah Standarisasi ===")
print(X.head(), "\n")
```

```python
print("=== Hasil Preprocessing (5 baris pertama) ===")
X.head()
```

```python
# 4
X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.2,
    stratify=y,
    random_state=0
)

print(f'Jumlah data asli : {X.shape[0]}')
print(f'Jumlah data latih: {X_train.shape[0]}')
print(f'Jumlah data uji  : {X_test.shape[0]}')

print('=========')
print(f'Proporsi label data asli:\n{pd.Series(y).value_counts(normalize=True)}\n')
print(f'Proporsi label data latih:\n{pd.Series(y_train).value_counts(normalize=True)}\n')
print(f'Proporsi label data uji:\n{pd.Series(y_test).value_counts(normalize=True)}')
```
